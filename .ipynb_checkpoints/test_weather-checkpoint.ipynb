```python
# 🔑 REMPLACEZ PAR VOTRE VRAIE CLÉ API
API_KEY = "16dded2db3298c68ab32df47eeb74696"

def test_weather_api_simple():
    """Test basique de l'API météo"""
    
    print("🌤️  Test de l'API OpenWeatherMap")
    print("=" * 40)
    
    # Ville de test
    city = "Paris"
    
    # URL de l'API
    url = f"http://api.openweathermap.org/data/2.5/weather"
    
    # Paramètres de la requête
    params = {
        'q': city,
        'appid': API_KEY,
        'units': 'metric',  # Pour avoir les températures en Celsius
        'lang': 'fr'        # Descriptions en français
    }
    
    try:
        # Envoi de la requête
        print(f"📍 Test pour la ville : {city}")
        print(f"🔗 URL : {url}")
        print("⏳ Envoi de la requête...")
        
        response = requests.get(url, params=params, timeout=10)
        
        # Vérification du statut
        print(f"📊 Code de réponse : {response.status_code}")
        
        if response.status_code == 200:
            # Succès !
            data = response.json()
            
            print("✅ API FONCTIONNE PARFAITEMENT !")
            print("\n📋 Données reçues :")
            print("-" * 30)
            
            # Affichage des données principales
            temperature = data['main']['temp']
            description = data['weather'][0]['description']
            humidity = data['main']['humidity']
            city_name = data['name']
            
            print(f"🏙️  Ville : {city_name}")
            print(f"🌡️  Température : {temperature}°C")
            print(f"☁️  Conditions : {description}")
            print(f"💧 Humidité : {humidity}%")
            
            return True
            
        elif response.status_code == 401:
            print("❌ ERREUR : Clé API invalide ou non activée")
            print("💡 Solutions :")
            print("   1. Vérifiez que vous avez copié la bonne clé")
            print("   2. Attendez 10-15 minutes (activation en cours)")
            print("   3. Vérifiez votre compte OpenWeatherMap")
            return False
            
        elif response.status_code == 404:
            print("❌ ERREUR : Ville non trouvée")
            return False
            
        else:
            print(f"❌ ERREUR : Code {response.status_code}")
            print(f"📄 Réponse : {response.text}")
            return False
            
    except requests.exceptions.Timeout:
        print("❌ ERREUR : Timeout - Vérifiez votre connexion internet")
        return False
        
    except requests.exceptions.ConnectionError:
        print("❌ ERREUR : Problème de connexion internet")
        return False
        
    except Exception as e:
        print(f"❌ ERREUR INATTENDUE : {e}")
        return False
```
