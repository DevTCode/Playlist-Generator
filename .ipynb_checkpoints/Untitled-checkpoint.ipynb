```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.ensemble import RandomForestRegressor
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials

class MusicRecommendationSystem:
    def __init__(self, spotify_client_id=None, spotify_client_secret=None):
        """
        Initialise le système de recommandation
        """
        self.df = None
        self.scaler = StandardScaler()
        self.label_encoders = {}
        self.model = RandomForestRegressor(n_estimators=100, random_state=42)
        
        # Configuration Spotify (optionnelle)
        if spotify_client_id and spotify_client_secret:
            auth_manager = SpotifyClientCredentials(
                client_id=spotify_client_id,
                client_secret=spotify_client_secret
            )
            self.sp = spotipy.Spotify(auth_manager=auth_manager)
        else:
            self.sp = None
    
    def load_data(self, csv_path):
        """
        Charge et préprocesse le dataset
        """
        self.df = pd.read_csv(csv_path)
        self._preprocess_data()
        print(f"Dataset chargé : {len(self.df)} morceaux")
    
    def _preprocess_data(self):
        """
        Préprocesse les données : encodage, normalisation
        """
        # Colonnes contextuelles à encoder
        context_columns = ['contexte_temporel', 'contexte_meteo', 'contexte_activite', 
                          'contexte_emotionnel', 'contexte_social', 'contexte_lieu', 'genre']
        
        # Encodage des variables catégorielles
        for col in context_columns:
            if col in self.df.columns:
                le = LabelEncoder()
                self.df[col + '_encoded'] = le.fit_transform(self.df[col].fillna('unknown'))
                self.label_encoders[col] = le
        
        # Colonnes numériques pour la similarité
        self.audio_features = ['bpm_estime', 'energie', 'valence', 'danceability', 
                              'acousticness', 'instrumentalness', 'speechiness', 'liveness']
        
        # Normalisation des features audio
        audio_data = self.df[self.audio_features].fillna(0)
        self.df[self.audio_features] = self.scaler.fit_transform(audio_data)
    
    def calculate_context_score(self, track, user_context):
        """
        Calcule un score de pertinence basé sur le contexte utilisateur
        """
        score = 0.0
        
        # Correspondance exacte des contextes (poids fort)
        context_matches = {
            'contexte_temporel': 0.2,
            'contexte_meteo': 0.15,
            'contexte_activite': 0.25,
            'contexte_emotionnel': 0.25,
            'contexte_social': 0.1,
            'contexte_lieu': 0.15
        }
        
        for context_type, weight in context_matches.items():
            if context_type in user_context:
                user_value = user_context[context_type]
                track_value = track.get(context_type, '')
                
                if user_value == track_value:
                    score += weight
                elif user_value == 'any' or track_value == 'any':
                    score += weight * 0.5  # Correspondance partielle
        
        return min(score, 1.0)  # Score max = 1
    
    def calculate_audio_similarity(self, track, target_features):
        """
        Calcule la similarité audio entre un morceau et des caractéristiques cibles
        """
        track_features = [track[feat] for feat in self.audio_features]
        similarity = cosine_similarity([track_features], [target_features])[0][0]
        return similarity
    
    def get_target_audio_features(self, user_context):
        """
        Détermine les caractéristiques audio cibles selon le contexte
        """
        # Profils audio par contexte émotionnel
        emotion_profiles = {
            'joyeux': {'energie': 0.8, 'valence': 0.9, 'danceability': 0.7},
            'calme': {'energie': 0.3, 'valence': 0.6, 'danceability': 0.4},
            'energique': {'energie': 0.9, 'valence': 0.7, 'danceability': 0.8},
            'melancolique': {'energie': 0.2, 'valence': 0.3, 'danceability': 0.3},
            'motive': {'energie': 0.75, 'valence': 0.8, 'danceability': 0.6}
        }
        
        # Profils par activité
        activity_profiles = {
            'sport': {'energie': 0.9, 'bpm_estime': 1.0, 'danceability': 0.8},
            'travail': {'energie': 0.5, 'instrumentalness': 0.7, 'speechiness': 0.2},
            'detente': {'energie': 0.3, 'acousticness': 0.6, 'valence': 0.6},
            'cuisine': {'energie': 0.6, 'valence': 0.7, 'danceability': 0.6},
            'fete': {'energie': 0.9, 'valence': 0.9, 'danceability': 0.9}
        }
        
        # Combiner les profils
        target = {feat: 0.5 for feat in self.audio_features}  # Valeurs par défaut
        
        emotion = user_context.get('contexte_emotionnel', '')
        if emotion in emotion_profiles:
            for feat, value in emotion_profiles[emotion].items():
                if feat in target:
                    target[feat] = value
        
        activity = user_context.get('contexte_activite', '')
        if activity in activity_profiles:
            for feat, value in activity_profiles[activity].items():
                if feat in target:
                    target[feat] = (target[feat] + value) / 2  # Moyenne
        
        return [target[feat] for feat in self.audio_features]
    
    def recommend_tracks(self, user_context, limit=10, diversity_factor=0.3):
        """
        Recommande des morceaux basés sur le contexte utilisateur
        
        Args:
            user_context: dict avec les contextes utilisateur
            limit: nombre de recommandations
            diversity_factor: facteur de diversité (0-1)
        """
        if self.df is None:
            raise ValueError("Aucun dataset chargé. Utilisez load_data() d'abord.")
        
        recommendations = []
        target_audio = self.get_target_audio_features(user_context)
        
        # Calculer les scores pour chaque morceau
        for idx, track in self.df.iterrows():
            # Score contextuel
            context_score = self.calculate_context_score(track, user_context)
            
            # Score de similarité audio
            audio_score = self.calculate_audio_similarity(track, target_audio)
            
            # Score final combiné
            final_score = (context_score * 0.6) + (audio_score * 0.4)
            
            recommendations.append({
                'index': idx,
                'id': track['id'],
                'artiste': track['artiste'],
                'titre': track['titre'],
                'genre': track['genre'],
                'context_score': context_score,
                'audio_score': audio_score,
                'final_score': final_score,
                'track_data': track
            })
        
        # Trier par score final
        recommendations.sort(key=lambda x: x['final_score'], reverse=True)
        
        # Appliquer la diversité
        if diversity_factor > 0:
            recommendations = self._apply_diversity(recommendations, diversity_factor)
        
        return recommendations[:limit]
    
    def _apply_diversity(self, recommendations, diversity_factor):
        """
        Applique un facteur de diversité pour éviter les recommandations trop similaires
        """
        diverse_recs = []
        seen_artists = set()
        seen_genres = set()
        
        for rec in recommendations:
            artist = rec['artiste']
            genre = rec['genre']
            
            # Pénalité pour artiste/genre déjà vus
            penalty = 0
            if artist in seen_artists:
                penalty += diversity_factor * 0.5
            if genre in seen_genres:
                penalty += diversity_factor * 0.3
            
            rec['final_score'] -= penalty
            diverse_recs.append(rec)
            
            seen_artists.add(artist)
            seen_genres.add(genre)
        
        # Re-trier après application de la diversité
        diverse_recs.sort(key=lambda x: x['final_score'], reverse=True)
        return diverse_recs
    
    def get_spotify_recommendations(self, seed_tracks, user_context, limit=10):
        """
        Utilise l'API Spotify pour obtenir des recommandations supplémentaires
        """
        if not self.sp:
            print("API Spotify non configurée")
            return []
        
        try:
            # Paramètres basés sur le contexte
            target_params = {}
            
            activity = user_context.get('contexte_activite', '')
            emotion = user_context.get('contexte_emotionnel', '')
            
            if activity == 'sport':
                target_params.update({
                    'target_energy': 0.8,
                    'target_danceability': 0.7,
                    'min_tempo': 120
                })
            elif activity == 'detente':
                target_params.update({
                    'target_energy': 0.3,
                    'target_valence': 0.6,
                    'max_tempo': 100
                })
            
            if emotion == 'joyeux':
                target_params.update({
                    'target_valence': 0.8,
                    'target_energy': 0.7
                })
            elif emotion == 'melancolique':
                target_params.update({
                    'target_valence': 0.3,
                    'target_energy': 0.4
                })
            
            # Obtenir les recommandations Spotify
            results = self.sp.recommendations(
                seed_tracks=seed_tracks[:5],  # Max 5 seeds
                limit=limit,
                **target_params
            )
            
            spotify_recs = []
            for track in results['tracks']:
                spotify_recs.append({
                    'id': track['id'],
                    'titre': track['name'],
                    'artiste': ', '.join([artist['name'] for artist in track['artists']]),
                    'url': track['external_urls']['spotify'],
                    'image': track['album']['images'][0]['url'] if track['album']['images'] else '',
                    'source': 'spotify'
                })
            
            return spotify_recs
            
        except Exception as e:
            print(f"Erreur API Spotify: {e}")
            return []
    
    def create_playlist(self, user_context, limit=15, include_spotify=True):
        """
        Crée une playlist complète combinant dataset local et Spotify
        """
        # Recommandations depuis le dataset local
        local_recs = self.recommend_tracks(user_context, limit=limit//2)
        
        playlist = []
        
        # Ajouter les recommandations locales
        for rec in local_recs:
            playlist.append({
                'titre': rec['titre'],
                'artiste': rec['artiste'],
                'genre': rec['genre'],
                'score': rec['final_score'],
                'source': 'local'
            })
        
        # Ajouter les recommandations Spotify si disponible
        if include_spotify and self.sp and local_recs:
            # Utiliser les IDs des meilleures recommandations locales comme seeds
            seed_tracks = [rec['id'] for rec in local_recs[:3] if rec['id']]
            
            if seed_tracks:
                spotify_recs = self.get_spotify_recommendations(
                    seed_tracks, user_context, limit=limit//2
                )
                playlist.extend(spotify_recs)
        
        return playlist[:limit]

# Exemple d'utilisation
def exemple_utilisation():
    """
    Exemple d'utilisation du système de recommandation
    """
    # Initialiser le système
    recommender = MusicRecommendationSystem(
        spotify_client_id='aa2a7bd2dcc84f19bb479185b3224ff6',  # Optionnel
        spotify_client_secret='49802f5d2d754ea8a78f01432b062d8b'  # Optionnel
    )
    
    # Charger les données
    recommender.load_data('votre_dataset.csv')
    
    # Définir le contexte utilisateur
    user_context = {
        'contexte_temporel': 'matin',
        'contexte_meteo': 'ensoleille',
        'contexte_activite': 'sport',
        'contexte_emotionnel': 'energique',
        'contexte_social': 'seul',
        'contexte_lieu': 'exterieur'
    }
    
    # Obtenir des recommandations
    recommendations = recommender.recommend_tracks(user_context, limit=10)
    
    # Afficher les résultats
    print("🎵 Recommandations personnalisées:")
    for i, rec in enumerate(recommendations, 1):
        print(f"{i}. {rec['titre']} - {rec['artiste']}")
        print(f"   Genre: {rec['genre']} | Score: {rec['final_score']:.3f}")
        print(f"   Context: {rec['context_score']:.3f} | Audio: {rec['audio_score']:.3f}")
        print()
    
    # Créer une playlist complète
    playlist = recommender.create_playlist(user_context, limit=15)
    
    print("🎧 Playlist générée:")
    for i, track in enumerate(playlist, 1):
        print(f"{i}. {track['titre']} - {track['artiste']} ({track['source']})")

if __name__ == "__main__":
    exemple_utilisation()
```
