### Importation des bibliothèques

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import warnings
warnings.filterwarnings('ignore')
```

### Configuration pour l'affichage

```python
print("\n📁 1. CHARGEMENT DES DONNÉES")
print("-" * 30)

# Chargement du dataset
df = pd.read_csv('../tracks_dataset.csv')

print(f"✅ Dataset chargé avec succès")
print(f"📊 Dimensions : {df.shape[0]} lignes × {df.shape[1]} colonnes")
print(f"💾 Taille mémoire : {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# Affichage des informations générales
print("\n📋 INFORMATIONS GÉNÉRALES :")
print(df.info())

print("\n🔍 APERÇU DES DONNÉES :")
print(df.head())

print("\n📈 STATISTIQUES DESCRIPTIVES :")
print(df.describe())
```

### Valeurs Manquantes

```python
print("\n🔍 2. VÉRIFICATION DES VALEURS MANQUANTES")
print("-" * 45)

# Calcul des valeurs manquantes
missing_data = df.isnull().sum()
missing_percent = (missing_data / len(df)) * 100

missing_df = pd.DataFrame({
    'Colonne': missing_data.index,
    'Valeurs_Manquantes': missing_data.values,
    'Pourcentage': missing_percent.values
})

missing_df = missing_df[missing_df['Valeurs_Manquantes'] > 0].sort_values('Valeurs_Manquantes', ascending=False)

if len(missing_df) > 0:
    print("⚠️  Colonnes avec valeurs manquantes :")
    print(missing_df.to_string(index=False))
    
    # Visualisation des valeurs manquantes
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    missing_df.plot(x='Colonne', y='Valeurs_Manquantes', kind='bar', ax=plt.gca())
    plt.title('Nombre de valeurs manquantes par colonne')
    plt.xticks(rotation=45)
    plt.tight_layout()
    
    plt.subplot(1, 2, 2)
    missing_df.plot(x='Colonne', y='Pourcentage', kind='bar', ax=plt.gca(), color='orange')
    plt.title('Pourcentage de valeurs manquantes')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("✅ Aucune valeur manquante détectée dans le dataset !")
```

### Identification des doublons

```python
print("\n🔄 3. IDENTIFICATION DES DOUBLONS")
print("-" * 35)

# Vérification des doublons complets
duplicates_total = df.duplicated().sum()
print(f"🔍 Doublons complets : {duplicates_total}")

# Vérification des doublons par ID (clé unique)
if 'id' in df.columns:
    duplicate_ids = df.duplicated(subset=['id']).sum()
    print(f"🆔 Doublons par ID : {duplicate_ids}")
    
    if duplicate_ids > 0:
        print("⚠️  IDs dupliqués trouvés :")
        duplicated_id_rows = df[df.duplicated(subset=['id'], keep=False)].sort_values('id')
        print(duplicated_id_rows[['id', 'artiste', 'titre']])

# Vérification des doublons par combinaison artiste + titre
if 'artiste' in df.columns and 'titre' in df.columns:
    duplicate_songs = df.duplicated(subset=['artiste', 'titre']).sum()
    print(f"🎵 Doublons artiste + titre : {duplicate_songs}")
    
    if duplicate_songs > 0:
        print("⚠️  Chansons dupliquées (artiste + titre) :")
        duplicated_songs = df[df.duplicated(subset=['artiste', 'titre'], keep=False)].sort_values(['artiste', 'titre'])
        print(duplicated_songs[['artiste', 'titre', 'genre']].head(10))
```

### Distribution des genres musicaux

```python
print("\n🎼 4. ANALYSE DES GENRES MUSICAUX")
print("-" * 35)

if 'genre' in df.columns:
    # Distribution des genres
    genre_counts = df['genre'].value_counts()
    print(f"📊 Nombre de genres uniques : {len(genre_counts)}")
    print(f"🎵 Genre le plus fréquent : {genre_counts.index[0]} ({genre_counts.iloc[0]} tracks)")
    print(f"🎵 Genre le moins fréquent : {genre_counts.index[-1]} ({genre_counts.iloc[-1]} tracks)")
    
    print("\n📈 Distribution des 15 genres principaux :")
    print(genre_counts.head(15))
    
    # Visualisation des genres
    plt.figure(figsize=(15, 10))
    
    # Top 15 genres
    plt.subplot(2, 2, 1)
    genre_counts.head(15).plot(kind='bar')
    plt.title('Top 15 Genres Musicaux')
    plt.xlabel('Genre')
    plt.ylabel('Nombre de tracks')
    plt.xticks(rotation=45)
    
    # Distribution en camembert pour top 10
    plt.subplot(2, 2, 2)
    top_10_genres = genre_counts.head(10)
    autres = genre_counts.iloc[10:].sum()
    if autres > 0:
        top_10_genres['Autres'] = autres
    
    plt.pie(top_10_genres.values, labels=top_10_genres.index, autopct='%1.1f%%')
    plt.title('Répartition des Genres (Top 10 + Autres)')
    
    # Distribution logarithmique
    plt.subplot(2, 2, 3)
    plt.hist(genre_counts.values, bins=20, edgecolor='black')
    plt.title('Distribution du nombre de tracks par genre')
    plt.xlabel('Nombre de tracks')
    plt.ylabel('Nombre de genres')
    plt.yscale('log')
    
    # Genres rares (moins de 5 tracks)
    plt.subplot(2, 2, 4)
    rare_genres = genre_counts[genre_counts <= 5]
    if len(rare_genres) > 0:
        rare_genres.plot(kind='bar')
        plt.title(f'Genres rares (≤ 5 tracks) - {len(rare_genres)} genres')
        plt.xticks(rotation=45)
    else:
        plt.text(0.5, 0.5, 'Aucun genre rare détecté', ha='center', va='center', transform=plt.gca().transAxes)
        plt.title('Genres rares')
    
    plt.tight_layout()
    plt.show()
    
    # Analyse des genres problématiques
    print(f"\n🔍 Genres avec peu de données (≤ 5 tracks) : {len(rare_genres)}")
    if len(rare_genres) > 0:
        print("Genres rares à considérer pour regroupement :")
        print(rare_genres.to_string())
```

### Analyse des contextes

```python
print("\n🌍 5. ANALYSE DES DIFFÉRENTS CONTEXTES")
print("-" * 40)

# Définition des colonnes de contexte
context_columns = [col for col in df.columns if 'contexte_' in col]
print(f"📋 Colonnes de contexte identifiées : {context_columns}")

# Analyse de chaque contexte
for context_col in context_columns:
    print(f"\n📊 {context_col.upper().replace('CONTEXTE_', '')} :")
    print("-" * 25)
    
    context_counts = df[context_col].value_counts()
    print(f"🔢 Valeurs uniques : {len(context_counts)}")
    print(f"📈 Répartition :")
    
    for value, count in context_counts.head(10).items():
        percentage = (count / len(df)) * 100
        print(f"  • {value}: {count} ({percentage:.1f}%)")
    
    if len(context_counts) > 10:
        print(f"  • ... et {len(context_counts) - 10} autres valeurs")

# Visualisation des contextes principaux
if len(context_columns) > 0:
    n_contexts = len(context_columns)
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    axes = axes.flatten()
    
    for i, context_col in enumerate(context_columns[:6]):  # Limite à 6 contextes
        if i < len(axes):
            context_counts = df[context_col].value_counts()
            
            # Limiter à top 10 pour la lisibilité
            if len(context_counts) > 10:
                top_contexts = context_counts.head(10)
                autres = context_counts.iloc[10:].sum()
                if autres > 0:
                    top_contexts['Autres'] = autres
                context_counts = top_contexts
            
            context_counts.plot(kind='bar', ax=axes[i])
            axes[i].set_title(f'{context_col.replace("contexte_", "").title()}')
            axes[i].tick_params(axis='x', rotation=45)
    
    # Masquer les axes non utilisés
    for j in range(len(context_columns), len(axes)):
        axes[j].set_visible(False)
    
    plt.tight_layout()
    plt.show()
```

### CARACTÉRISTIQUES AUDIO - VÉRIFICATION DES PLAGES

```python
print("\n🎵 6. ANALYSE DES CARACTÉRISTIQUES AUDIO")
print("-" * 42)

# Colonnes de caractéristiques audio
audio_features = ['energie', 'valence', 'danceability', 'acousticness', 
                 'instrumentalness', 'speechiness', 'liveness', 'bpm_estime', 'notes']

# Vérification des plages de valeurs
print("📊 PLAGES DE VALEURS DES CARACTÉRISTIQUES AUDIO :")
print("-" * 50)

audio_stats = pd.DataFrame()
for feature in audio_features:
    if feature in df.columns:
        stats = {
            'Minimum': df[feature].min(),
            'Maximum': df[feature].max(),
            'Moyenne': df[feature].mean(),
            'Médiane': df[feature].median(),
            'Écart-type': df[feature].std(),
            'Valeurs_nulles': df[feature].isnull().sum()
        }
        audio_stats[feature] = stats

audio_stats = audio_stats.T
print(audio_stats.round(3))

# Identification des valeurs aberrantes
print("\n⚠️  DÉTECTION DE VALEURS ABERRANTES :")
print("-" * 35)

# Vérification des plages attendues (0-1 pour la plupart)
expected_ranges = {
    'energie': (0, 1),
    'valence': (0, 1),
    'danceability': (0, 1),
    'acousticness': (0, 1),
    'instrumentalness': (0, 1),
    'speechiness': (0, 1),
    'liveness': (0, 1),
    'bpm_estime': (50, 200),  # BPM typique
    'notes': (0, 10)  # Notes sur 10
}

for feature, (min_val, max_val) in expected_ranges.items():
    if feature in df.columns:
        outliers_low = (df[feature] < min_val).sum()
        outliers_high = (df[feature] > max_val).sum()
        total_outliers = outliers_low + outliers_high
        
        if total_outliers > 0:
            print(f"🚨 {feature}: {total_outliers} valeurs aberrantes")
            if outliers_low > 0:
                print(f"   • {outliers_low} valeurs < {min_val}")
            if outliers_high > 0:
                print(f"   • {outliers_high} valeurs > {max_val}")
        else:
            print(f"✅ {feature}: Toutes les valeurs dans la plage attendue")

# Visualisation des distributions
if len([f for f in audio_features if f in df.columns]) > 0:
    plt.figure(figsize=(15, 12))
    
    available_features = [f for f in audio_features if f in df.columns]
    n_features = len(available_features)
    
    for i, feature in enumerate(available_features):
        plt.subplot(3, 3, i + 1)
        
        # Histogramme avec courbe de densité
        df[feature].hist(bins=30, alpha=0.7, density=True)
        df[feature].plot(kind='kde', secondary_y=True)
        plt.title(f'Distribution - {feature}')
        plt.xlabel(feature)
        
        # Ajout de lignes pour les valeurs aberrantes si applicable
        if feature in expected_ranges:
            min_val, max_val = expected_ranges[feature]
            plt.axvline(min_val, color='red', linestyle='--', alpha=0.7, label=f'Min={min_val}')
            plt.axvline(max_val, color='red', linestyle='--', alpha=0.7, label=f'Max={max_val}')
    
    plt.tight_layout()
    plt.show()
```

### RÉSUMÉ DE L'INSPECTION INITIALE

```python
print("\n📋 7. RÉSUMÉ DE L'INSPECTION INITIALE")
print("=" * 50)

print(f"✅ Dataset analysé : {df.shape[0]} tracks, {df.shape[1]} caractéristiques")
print(f"✅ Valeurs manquantes : {df.isnull().sum().sum()} au total")
print(f"✅ Doublons complets : {df.duplicated().sum()}")

if 'genre' in df.columns:
    print(f"✅ Genres musicaux : {df['genre'].nunique()} genres uniques")
    rare_genres_count = (df['genre'].value_counts() <= 5).sum()
    print(f"⚠️  Genres rares (≤ 5 tracks) : {rare_genres_count}")

print(f"✅ Contextes analysés : {len(context_columns)} types de contexte")
```
