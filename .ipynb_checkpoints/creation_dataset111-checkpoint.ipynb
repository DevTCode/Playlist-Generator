````python
# Guide de Création du Dataset - Générateur de Playlists Contextuel

## 1. Configuration des APIs

### 1.1 Spotify Web API
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
import pandas as pd
import requests
import json
import pylast
from datetime import datetime, timedelta
````

````python
# Configuration Spotify
client_credentials_manager = SpotifyClientCredentials(
    client_id="1a1fcf3039b54d78b8266d32451a75ba",
    client_secret="dfc45caedccf4e1a8ae363559c4d1f65"
)
sp = spotipy.Spotify(client_credentials_manager=client_credentials_manager)

# Fonction pour extraire les caractéristiques audio
def get_audio_features(track_ids):
    """Récupère les caractéristiques audio pour une liste de tracks"""
    features = sp.audio_features(track_ids)
    return features

# Fonction pour rechercher des playlists par contexte
def search_contextual_playlists(query, limit=50):
    """Recherche des playlists par mots-clés contextuels"""
    results = sp.search(q=query, type='playlist', limit=limit)
    return results['playlists']['items']
```

### 1.2 OpenWeatherMap API
# Configuration météo
WEATHER_API_KEY = "16dded2db3298c68ab32df47eeb74696"

def get_weather_data(city):
    """Récupère les données météo actuelles"""
    url = f"http://api.openweathermap.org/data/2.5/weather"
    params = {
        'q': city,
        'appid': WEATHER_API_KEY,
        'units': 'metric'
    }
    response = requests.get(url, params=params)
    return response.json()

def get_weather_history(lat, lon, start_date, end_date):
    """Récupère l'historique météo pour créer des corrélations"""
    # Utilisation de l'API historique d'OpenWeatherMap
    url = f"http://api.openweathermap.org/data/2.5/onecall/timemachine"
    weather_data = []
    
    current_date = start_date
    while current_date <= end_date:
        timestamp = int(current_date.timestamp())
        params = {
            'lat': lat,
            'lon': lon,
            'dt': timestamp,
            'appid': WEATHER_API_KEY,
            'units': 'metric'
        }
        response = requests.get(url, params=params)
        weather_data.append(response.json())
        current_date += timedelta(days=1)
    
    return weather_data

### 1.3 Last.fm API
# Configuration Last.fm
LASTFM_API_KEY = "78d22b18b58bde175a2e0e4118e392bb"
LASTFM_API_SECRET = "1f5bdc5bcee5746fdd8af7310c94d62c"

network = pylast.LastFMNetwork(
    api_key=LASTFM_API_KEY,
    api_secret=LASTFM_API_SECRET
)

def get_track_tags(artist, track):
    """Récupère les tags communautaires d'une track"""
    try:
        track_obj = network.get_track(artist, track)
        tags = track_obj.get_top_tags(limit=10)
        return [tag.item.name for tag in tags]
    except:
        return []

def get_similar_tracks(artist, track, limit=20):
    """Récupère des tracks similaires"""
    try:
        track_obj = network.get_track(artist, track)
        similar = track_obj.get_similar(limit=limit)
        return [(s.item.artist.name, s.item.title, s.match) for s in similar]
    except:
        return []

## 2. Collecte des Données Musicales

### 2.1 Extraction de Playlists Contextuelles
def collect_contextual_playlists():
    """Collecte des playlists pour différents contextes"""
    
    # Contextes à explorer
    contexts = {
        'weather': ['rainy day', 'sunny day', 'winter music', 'summer vibes'],
        'time': ['morning music', 'evening chill', 'night time', 'afternoon'],
        'activity': ['workout', 'study music', 'cooking', 'driving', 'relaxation'],
        'mood': ['happy', 'sad', 'energetic', 'calm', 'motivated'],
        'social': ['party', 'romantic', 'friends', 'alone time']
    }
    
    all_playlists = []
    
    for context_type, keywords in contexts.items():
        for keyword in keywords:
            playlists = search_contextual_playlists(keyword)
            for playlist in playlists:
                playlist_data = {
                    'context_type': context_type,
                    'context_keyword': keyword,
                    'playlist_id': playlist['id'],
                    'playlist_name': playlist['name'],
                    'playlist_description': playlist['description'],
                    'follower_count': playlist['followers']['total'],
                    'track_count': playlist['tracks']['total']
                }
                all_playlists.append(playlist_data)
    
    return pd.DataFrame(all_playlists)

def extract_tracks_from_playlists(playlist_df):
    """Extrait toutes les tracks des playlists collectées"""
    
    all_tracks = []
    
    for _, playlist in playlist_df.iterrows():
        try:
            # Récupération des tracks de la playlist
            results = sp.playlist_tracks(playlist['playlist_id'])
            tracks = results['items']
            
            # Pagination pour récupérer toutes les tracks
            while results['next']:
                results = sp.next(results)
                tracks.extend(results['items'])
            
            for item in tracks:
                if item['track'] and item['track']['id']:
                    track_data = {
                        'context_type': playlist['context_type'],
                        'context_keyword': playlist['context_keyword'],
                        'playlist_id': playlist['playlist_id'],
                        'track_id': item['track']['id'],
                        'track_name': item['track']['name'],
                        'artist_name': item['track']['artists'][0]['name'],
                        'album_name': item['track']['album']['name'],
                        'duration_ms': item['track']['duration_ms'],
                        'popularity': item['track']['popularity'],
                        'explicit': item['track']['explicit'],
                        'release_date': item['track']['album']['release_date']
                    }
                    all_tracks.append(track_data)
                    
        except Exception as e:
            print(f"Erreur pour playlist {playlist['playlist_id']}: {e}")
            continue
    
    return pd.DataFrame(all_tracks)

### 2.2 Enrichissement avec Caractéristiques Audio
def enrich_with_audio_features(tracks_df):
    """Ajoute les caractéristiques audio Spotify"""
    
    # Traitement par lots de 100 tracks max
    batch_size = 100
    enriched_tracks = []
    
    for i in range(0, len(tracks_df), batch_size):
        batch = tracks_df.iloc[i:i+batch_size]
        track_ids = batch['track_id'].tolist()
        
        # Récupération des features audio
        audio_features = get_audio_features(track_ids)
        
        for j, features in enumerate(audio_features):
            if features:  # Si les features sont disponibles
                track_data = batch.iloc[j].to_dict()
                track_data.update({
                    'danceability': features['danceability'],
                    'energy': features['energy'],
                    'key': features['key'],
                    'loudness': features['loudness'],
                    'mode': features['mode'],
                    'speechiness': features['speechiness'],
                    'acousticness': features['acousticness'],
                    'instrumentalness': features['instrumentalness'],
                    'liveness': features['liveness'],
                    'valence': features['valence'],
                    'tempo': features['tempo'],
                    'time_signature': features['time_signature']
                })
                enriched_tracks.append(track_data)
    
    return pd.DataFrame(enriched_tracks)

## 3. Collecte des Données Contextuelles

### 3.1 Données Météorologiques Historiques
def create_weather_music_dataset():
    """Crée un dataset météo-musique pour l'entraînement"""
    
    # Principales villes pour la diversité géographique
    cities = [
        ('Paris', 48.8566, 2.3522),
        ('Londres', 51.5074, -0.1278),
        ('New York', 40.7128, -74.0060),
        ('Tokyo', 35.6762, 139.6503),
        ('Sydney', -33.8688, 151.2093)
    ]
    
    # Période d'analyse (1 an)
    start_date = datetime.now() - timedelta(days=365)
    end_date = datetime.now()
    
    weather_data = []
    
    for city_name, lat, lon in cities:
        city_weather = get_weather_history(lat, lon, start_date, end_date)
        
        for day_data in city_weather:
            weather_entry = {
                'city': city_name,
                'date': datetime.fromtimestamp(day_data['current']['dt']),
                'temperature': day_data['current']['temp'],
                'feels_like': day_data['current']['feels_like'],
                'humidity': day_data['current']['humidity'],
                'pressure': day_data['current']['pressure'],
                'visibility': day_data['current'].get('visibility', 10000),
                'weather_main': day_data['current']['weather'][0]['main'],
                'weather_description': day_data['current']['weather'][0]['description'],
                'cloudiness': day_data['current']['clouds'],
                'wind_speed': day_data['current']['wind_speed']
            }
            weather_data.append(weather_entry)
    
    return pd.DataFrame(weather_data)

### 3.2 Données Temporelles et Circadiennes
def generate_temporal_patterns():
    """Génère des patterns temporels pour l'analyse"""
    
    temporal_data = []
    
    # Analyse par heure de la journée
    for hour in range(24):
        # Définition des contextes par heure
        if 6 <= hour < 10:
            period = 'morning'
            energy_preference = 0.6
            valence_preference = 0.7
        elif 10 <= hour < 14:
            period = 'midday'
            energy_preference = 0.7
            valence_preference = 0.8
        elif 14 <= hour < 18:
            period = 'afternoon'
            energy_preference = 0.5
            valence_preference = 0.6
        elif 18 <= hour < 22:
            period = 'evening'
            energy_preference = 0.4
            valence_preference = 0.5
        else:
            period = 'night'
            energy_preference = 0.3
            valence_preference = 0.4
        
        temporal_entry = {
            'hour': hour,
            'period': period,
            'energy_preference': energy_preference,
            'valence_preference': valence_preference,
            'tempo_min': 80 + (hour * 2),
            'tempo_max': 120 + (hour * 3)
        }
        temporal_data.append(temporal_entry)
    
    return pd.DataFrame(temporal_data)

## 4. Labellisation et Scoring Contextuel

### 4.1 Création des Scores de Pertinence
def create_contextual_scores(tracks_df, weather_df):
    """Crée des scores de pertinence contextuelle"""
    
    scored_tracks = []
    
    for _, track in tracks_df.iterrows():
        # Score météorologique
        weather_score = calculate_weather_score(track, weather_df)
        
        # Score temporel
        temporal_score = calculate_temporal_score(track)
        
        # Score d'activité
        activity_score = calculate_activity_score(track)
        
        # Score émotionnel
        mood_score = calculate_mood_score(track)
        
        track_scored = track.to_dict()
        track_scored.update({
            'weather_score': weather_score,
            'temporal_score': temporal_score,
            'activity_score': activity_score,
            'mood_score': mood_score,
            'overall_context_score': (weather_score + temporal_score + 
                                    activity_score + mood_score) / 4
        })
        scored_tracks.append(track_scored)
    
    return pd.DataFrame(scored_tracks)

def calculate_weather_score(track, weather_context):
    """Calcule le score de pertinence météorologique"""
    
    # Règles basées sur les caractéristiques audio
    if weather_context['weather_main'] == 'Rain':
        # Musique plus calme et mélancolique pour la pluie
        score = (1 - track['energy']) * 0.4 + (1 - track['valence']) * 0.3 + track['acousticness'] * 0.3
    elif weather_context['weather_main'] == 'Clear':
        # Musique énergique et positive pour le beau temps
        score = track['energy'] * 0.4 + track['valence'] * 0.4 + (1 - track['acousticness']) * 0.2
    elif weather_context['weather_main'] == 'Snow':
        # Musique douce et contemplative pour la neige
        score = track['acousticness'] * 0.4 + (track['valence'] * 0.5) * 0.3 + (1 - track['energy']) * 0.3
    else:
        # Score neutre pour les autres conditions
        score = 0.5
    
    return min(max(score, 0), 1)  # Normalisation entre 0 et 1

## 5. Nettoyage et Validation

### 5.1 Nettoyage des Données
def clean_dataset(df):
    """Nettoie et valide le dataset"""
    
    # Suppression des doublons
    df = df.drop_duplicates(subset=['track_id'])
    
    # Suppression des valeurs manquantes critiques
    df = df.dropna(subset=['track_id', 'energy', 'valence', 'danceability'])
    
    # Normalisation des caractéristiques audio (déjà normalisées par Spotify)
    audio_features = ['danceability', 'energy', 'speechiness', 'acousticness', 
                     'instrumentalness', 'liveness', 'valence']
    
    # Vérification que les valeurs sont bien entre 0 et 1
    for feature in audio_features:
        df[feature] = df[feature].clip(0, 1)
    
    # Gestion des outliers pour le tempo
    df['tempo'] = df['tempo'].clip(60, 200)
    
    # Encodage des variables catégorielles
    df = pd.get_dummies(df, columns=['context_type', 'context_keyword'], prefix=['ctx_type', 'ctx_kw'])
    
    return df

def validate_dataset(df):
    """Valide la qualité du dataset"""
    
    validation_report = {
        'total_tracks': len(df),
        'unique_tracks': df['track_id'].nunique(),
        'missing_values': df.isnull().sum().to_dict(),
        'context_distribution': df['context_type'].value_counts().to_dict() if 'context_type' in df.columns else {},
        'audio_features_stats': df[['energy', 'valence', 'danceability']].describe().to_dict()
    }
    
    return validation_report

## 6. Export et Sauvegarde

### 6.1 Sauvegarde du Dataset
```python
def save_dataset(df, filename='contextual_music_dataset.csv'):
    """Sauvegarde le dataset final"""
    
    # Sauvegarde CSV
    df.to_csv(filename, index=False, encoding='utf-8')
    
    # Sauvegarde Parquet pour de meilleures performances
    df.to_parquet(filename.replace('.csv', '.parquet'), index=False)
    
    # Création d'un fichier de métadonnées
    metadata = {
        'creation_date': datetime.now().isoformat(),
        'total_tracks': len(df),
        'features': list(df.columns),
        'contexts_covered': df['context_type'].unique().tolist() if 'context_type' in df.columns else [],
        'data_sources': ['Spotify Web API', 'Last.fm API', 'OpenWeatherMap API']
    }
    
    with open(filename.replace('.csv', '_metadata.json'), 'w') as f:
        json.dump(metadata, f, indent=2)
    
    print(f"Dataset sauvegardé : {filename}")
    print(f"Métadonnées : {filename.replace('.csv', '_metadata.json')}")

# Exemple d'utilisation complète
if __name__ == "__main__":
    # 1. Collecte des playlists contextuelles
    playlists_df = collect_contextual_playlists()
    
    # 2. Extraction des tracks
    tracks_df = extract_tracks_from_playlists(playlists_df)
    
    # 3. Enrichissement avec features audio
    enriched_tracks_df = enrich_with_audio_features(tracks_df)
    
    # 4. Collecte des données météo
    weather_df = create_weather_music_dataset()
    
    # 5. Scoring contextuel
    scored_tracks_df = create_contextual_scores(enriched_tracks_df, weather_df)
    
    # 6. Nettoyage
    clean_df = clean_dataset(scored_tracks_df)
    
    # 7. Validation
    validation = validate_dataset(clean_df)
    print("Validation du dataset :", validation)
    
    # 8. Sauvegarde
    save_dataset(clean_df)
````
